# Boolean Mask

```python
alcohol_under_13_boolean_mask = alcohol <13
# print(alcohol_under_13_boolean_mask)
```

```python
wine_data[alcohol_under_13_boolean_mask,:].shape #It filters the false value(under 13)
```

## To add more filter to the mask , just use AND operator or use OR operator, where you see fit
AND(&) --> Return True if all condition are true\n
OR(|) --> Return True if one condition is true


```python
magnesium_over_120_boolean_mask = wine_data[:,4] > 120
```

```python
# magnesium_over_120_boolean_mask
```

```python
wine_data[magnesium_over_120_boolean_mask].shape
```

```python
combined_mask = alcohol_under_13_boolean_mask & magnesium_over_120_boolean_mask 
```

```python
wine_data[combined_mask].shape
```

### np.where(condition_for_array , value_return_if_true, value_return_if_false)

```python
low_alcohol = np.where(alcohol < 12, "Low alcohol", "High alcohol")
```

```python
print(low_alcohol[alcohol< 12])
```

```python
#To nest the conditions

low_alcohol = np.where(alcohol < 12, "Low alcohol", np.where(alcohol > 16,"High alcohol", "Moderate alcohol"))
```

```python
print(low_alcohol.shape)
```

```python
array = np.sort(alcohol) # To Sort in acending order
# print(array)
```

```python
array = np.sort(alcohol)[::-1] #for decending order
# print(array)
```

```python
indices = np.argsort(alcohol) # Returns the indicies that can sort the array
# Can be used to sort other array in same order
alcohol[indices]
wine_data[indices] #both get sorted to same order
```

### Ex - if you have one rollno array and one names array, rollno are according to the names and are releated to only coloumn, if use argsort on the rollno array, you get the indices array which you apply to both rollno and names array

## Numpy arrays are homogeneous , which means you cannot have different datatypes of element in an array

```python
np.array([1,"2",3, 4.5]) #they will become same datatype
#least common denominator data type
# if atleast one string is present, all elements(integer, float, boolean,etc) will become string
```

```python
np.array([1.2, 3,4, 5])
# if atleast one float is present, all integer will become float
```
