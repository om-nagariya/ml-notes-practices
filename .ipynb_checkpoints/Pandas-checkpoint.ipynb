```python
 # df.map?
```

```python
# Takes function and lables the string returned by the function according to given coloumn in the function definition
```

```python
def map_ancient_category(years):
    if years >=20:
        return "Almost hit the sack"
    elif years >= 10:
        return "Late bloomers"
    elif years >= 5:
        return "Experience Novice"
    else:
        return "Noive"
```

```python
df['Ancientinity'] = df['Years of Experience'].map(map_ancient_category)
```
