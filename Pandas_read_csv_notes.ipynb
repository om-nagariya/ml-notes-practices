# Pandas `read_csv()` — Notes

`pd.read_csv()` loads CSV/delimited files into a **pandas DataFrame**. Its parameters control how the file is interpreted, cleaned, typed, and loaded.

### Core idea
**Read → Select → Clean → Type → Parse → Handle missing data → Handle large files**

````python
import pandas as pd

df = pd.read_csv("data.csv")
````

## 1. `pd.read_csv()`

**Purpose:** Read a CSV/delimited file into a DataFrame.

```python
df = pd.read_csv("data.csv")
df = pd.read_csv("data/movies.csv")
```

Can read from a local path or URL and can control separators, headers, missing values, dates, data types, and file size.

## 2. `requests.get()`

**Purpose:** Download file content from a URL before passing it to pandas.

```python
import requests
from io import StringIO

response = requests.get(url)
df = pd.read_csv(StringIO(response.text))
```

## Important `read_csv()` Parameters

| Parameter | Remember it as | Main use |
|---|---|---|
| `sep` | separator | Define delimiter |
| `names` | names | Manually provide column names |
| `index_col` | index | Set a column as index |
| `header` | header row | Choose header row |
| `usecols` | use columns | Load only selected columns |
| `skiprows` | skip rows | Ignore selected rows |
| `nrows` | number of rows | Read limited rows |
| `encoding` | character encoding | Handle special/non-English characters |
| `on_bad_lines` | bad rows | Handle malformed rows |
| `dtype` | data type | Force column types |
| `parse_dates` | dates | Convert to datetime |
| `converters` | custom function | Transform values while reading |
| `na_values` | missing values | Define custom missing markers |
| `chunksize` | batches | Read large files in chunks |

## 3. `sep` ⭐

**Defines the column separator.** Default is comma.

```python
df = pd.read_csv("data.csv", sep=",")
df = pd.read_csv("data.tsv", sep="\t")
```

**Remember:** Wrong separator → columns may not split correctly.

## 4. `names` ⭐

**Manually provides column names**, useful when the file has no header.

```python
columns = ["id", "movie", "year", "rating", "votes"]
df = pd.read_csv("movies.csv", names=columns)
```

## 5. `index_col` ⭐

**Sets a column as the DataFrame index.**

```python
df = pd.read_csv("data.csv", index_col="movie_id")
df = pd.read_csv("data.csv", index_col=0)
```

## 6. `header` ⭐

**Specifies which row contains column names.**

```python
df = pd.read_csv("data.csv", header=1)
```

Useful when the actual column names are not on the first row.

## 7. `usecols` ⭐⭐⭐

**Loads only required columns.** Helps reduce memory and simplify the DataFrame.

```python
df = pd.read_csv(
    "students.csv",
    usecols=["enrolment_id", "gender", "education_level"]
)
```

Can also use positions:

```python
df = pd.read_csv("data.csv", usecols=[0, 2, 4])
```

## 8. Selecting One Column

Selecting a single column from the resulting DataFrame gives a **Series**.

```python
series = pd.read_csv(
    "students.csv",
    usecols=["gender"]
)["gender"]
```

**Series = 1D** | **DataFrame = 2D**

## 9. `skiprows` ⭐

**Skips selected rows while reading.**

```python
df = pd.read_csv("data.csv", skiprows=[1, 2])
```

Can also use a function for custom rules:

```python
df = pd.read_csv(
    "data.csv",
    skiprows=lambda row_number: row_number % 2 == 0
)
```

## 10. `nrows` ⭐

**Reads only a specified number of rows.**

```python
df = pd.read_csv("large_file.csv", nrows=1000)
```

Useful for quick inspection, testing, or avoiding loading the entire file.

## 11. `encoding` ⭐

**Specifies character encoding.** Useful for special symbols, emojis, and non-English text.

```python
df = pd.read_csv("data.csv", encoding="utf-8")
df = pd.read_csv("data.csv", encoding="ISO-8859-1")
```

**Remember:** If you get a decoding error, check the file's actual encoding.

## 12. `on_bad_lines` ⭐

**Handles malformed CSV rows.**

```python
df = pd.read_csv("data.csv", on_bad_lines="skip")
```

The video mentions the older `error_bad_lines=False`; in newer pandas, use `on_bad_lines="skip"`.

## 13. `dtype` ⭐⭐⭐

**Controls the data type of columns during import.**

```python
df = pd.read_csv(
    "data.csv",
    dtype={"target": "int32", "age": "int8"}
)
```

Useful for correct types and memory optimization.

## 14. `parse_dates` ⭐⭐⭐

**Converts date columns into datetime values.**

```python
df = pd.read_csv("matches.csv", parse_dates=["date"])
```

Then datetime operations become possible:

```python
df["month"] = df["date"].dt.month
df["day"] = df["date"].dt.day
```

## 15. Combining Date Columns

Combine separate year/month/day columns into one date column.

```python
df = pd.read_csv(
    "data.csv",
    parse_dates={"date": ["year", "month", "day"]}
)
```

## 16. `converters` ⭐⭐⭐

**Applies a custom function to values in a column while reading.**

```python
def shorten_team_name(team):
    if team == "Royal Challengers Bangalore":
        return "RCB"
    return team

df = pd.read_csv(
    "matches.csv",
    converters={"team1": shorten_team_name}
)
```

Can transform multiple columns with different or shared functions.

## 17. `na_values` ⭐⭐⭐

**Defines custom values that should be treated as missing (`NaN`).**

```python
df = pd.read_csv(
    "students.csv",
    na_values=["-", "NA", "unknown"]
)
```

Useful when datasets use placeholders such as `-`, `NA`, or `unknown` for missing data.

## 18. `chunksize` ⭐⭐⭐

**Reads a large file in smaller DataFrame chunks instead of loading everything into RAM.**

```python
chunks = pd.read_csv("large_file.csv", chunksize=5000)

for chunk in chunks:
    print(chunk.shape)
```

Useful for very large datasets and memory-efficient processing.

# Combined Example

```python
import pandas as pd

def shorten_team_name(team):
    mapping = {
        "Royal Challengers Bangalore": "RCB",
        "Mumbai Indians": "MI",
        "Rajasthan Royals": "RR"
    }
    return mapping.get(team, team)

df = pd.read_csv(
    "matches.csv",
    sep=",",
    usecols=["date", "team1", "team2", "target"],
    parse_dates=["date"],
    dtype={"target": "int32"},
    converters={
        "team1": shorten_team_name,
        "team2": shorten_team_name
    },
    na_values=["-", "NA", "unknown"],
    on_bad_lines="skip"
)
```
